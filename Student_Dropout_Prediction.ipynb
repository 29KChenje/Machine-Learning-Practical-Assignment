```python
import pandas as pd

df = pd.read_csv("zimbabwe_student_dropout (1).csv")

# Check data
df.head()
```

```python
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
```

```python
df.isnull().sum()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.countplot(x='dropout', data=df)
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_cols:
    if col != 'dropout':
        sns.boxplot(x='dropout', y=col, data=df)
        plt.show()
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
df = df.dropna()
```

```python
df = pd.get_dummies(df, drop_first=True)
```

```python
target = 'dropout'  # change if needed

X = df.drop(target, axis=1)
y = df[target]
```

```python
print("After preprocessing:")
print(df.head())

print("\nDataset shape:", df.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Example: combine first 2 numeric columns
cols = X.columns

if len(cols) >= 2:
    df['new_feature_1'] = df[cols[0]] * df[cols[1]]
    df['new_feature_2'] = df[cols[0]] / (df[cols[1]] + 1)
```

```python
from sklearn.model_selection import train_test_split

X = df.drop(target, axis=1)
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression

model1 = LogisticRegression(max_iter=1000)
model1.fit(X_train, y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier

model2 = DecisionTreeClassifier()
model2.fit(X_train, y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier

model3 = RandomForestClassifier()
model3.fit(X_train, y_train)
```

```python
print("New features added:")
print(df[['new_feature_1', 'new_feature_2']].head())
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

models = {
    "Logistic": model1,
    "Decision Tree": model2,
    "Random Forest": model3
}

for name, model in models.items():
    pred = model.predict(X_test)
    print(f"\n{name}")
    print("Accuracy:", accuracy_score(y_test, pred))
    print(confusion_matrix(y_test, pred))
    print(classification_report(y_test, pred))
```
